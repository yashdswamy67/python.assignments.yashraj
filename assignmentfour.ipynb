```c++
import numpy as np

print("--- Matrix Addition ---")

# Define two 2x2 matrices using standard Python lists
matrix_A = [[1, 2], ]

matrix_B = [[5, 6], ]

# 1. Addition using standard Python Lists (Nested Loops)
list_result = [[0, 0], [0, 0]]
for i in range(len(matrix_A)):
    for j in range(len(matrix_A[0])):
        list_result[i][j] = matrix_A[i][j] + matrix_B[i][j]

print("Result using standard Python Lists:")
print(list_result)

# 2. Addition using NumPy Arrays (Optimized element-wise operations)
array_A = np.array(matrix_A)
array_B = np.array(matrix_B)
numpy_result = array_A + array_B

print("\nResult using NumPy Arrays:")
print(numpy_result)
```
