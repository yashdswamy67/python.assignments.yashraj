```c++
# --- LIST OPERATIONS ---
print("--- List Operations ---")
my_list = [10, 20, 30, 40]
my_list.append(50)          # Method 1: append()
my_list.insert(1, 15)       # Method 2: insert()
my_list.remove(30)          # Method 3: remove()
popped_val = my_list.pop()  # Method 4: pop()
my_list.sort(reverse=True)  # Method 5: sort()
print(f"Final List: {my_list}\n")

# --- TUPLE OPERATIONS ---
print("--- Tuple Operations ---")
my_tuple = (1, 2, 3, 2, 4, 2)
print(f"Length of tuple: {len(my_tuple)}")       # Method 1: len()
print(f"Count of element 2: {my_tuple.count(2)}") # Method 2: count()
print(f"Index of element 4: {my_tuple.index(4)}") # Method 3: index()
sliced_tuple = my_tuple[1:4]                      # Method 4: Slicing
min_val = min(my_tuple)                           # Method 5: min()
print(f"Sliced Tuple: {sliced_tuple}, Min Value: {min_val}\n")

# --- DICTIONARY OPERATIONS ---
print("--- Dictionary Operations ---")
student = {"name": "Amit", "roll_no": 101, "branch": "CS"}
student["email"] = "amit@mitwpu.edu.in"          # Method 1: Adding a key
student.update({"age": 20})                      # Method 2: update()
removed_branch = student.pop("branch")           # Method 3: pop()
keys = list(student.keys())                      # Method 4: keys()
values = list(student.values())                  # Method 5: values()
print(f"Final Dictionary: {student}")
```
