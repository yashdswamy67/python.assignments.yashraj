```c++
def check_right_angled_triangle(a, b, c):
    """
    Function to check if sides a, b, c form a right-angled triangle.
    Uses the Pythagorean theorem:
    square of hypotenuse = sum of squares of other two sides.
    """

    # Sort the sides so the largest side is the hypotenuse
    sides = sorted([a, b, c])

    # Pythagorean theorem
    if sides[0] ** 2 + sides[1] ** 2 == sides[2] ** 2:
        return True
    else:
        return False


# Main Program / Function Calling
print("--- Right-Angled Triangle Checker ---")

side1 = float(input("Enter length of side 1: "))
side2 = float(input("Enter length of side 2: "))
side3 = float(input("Enter length of side 3: "))

# Function call and output
if check_right_angled_triangle(side1, side2, side3):
    print("\nResult: The given sides FORM a right-angled triangle.")
else:
    print("\nResult: The given sides DO NOT FORM a right-angled triangle.")
```
